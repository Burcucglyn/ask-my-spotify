## 1-start with the usual drill import!


```python
import json
from pathlib import Path
import pandas as pd 
```

## 2-describe the paths to make sure that they aren't hard coded

```python
REPO_ROOT = Path.cwd().parent #nb is lives in the nb go up.
RAW = REPO_ROOT/"data"/"raw" #well it won't be raw after this...
```

## 3-well lets sort this bad raw jsons..

```python
#telling it find the newest recently played entitiy and load into python
#latest= find the newest (aka-the most novel) file pls, sort it for me, join a folder/file onto this path and make sure glob finds the absulute match,
#since youre the star take every thin ends with .json, and grab the last item from the sorted list
latest = sorted((RAW /"recently_played").glob("*.json"))[-1]

#print the latest finding with their name pls
print("loading:", latest.name)

#create read and load
data= json.loads(latest.read_text())
```

## 4-time to create dataframe..

```python
#dataframe name, pandas read the nested json bootstaps lives under the items key only
#check nested entity.. tell me the shape 
df = pd.json_normalize(data["items"])
df.shape
```
